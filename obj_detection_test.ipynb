# Webcam Display
### Code Adapted From: https://www.geeksforgeeks.org/python/python-opencv-capture-video-from-camera/

```python
cam = cv2.VideoCapture(0)

frame_width = int(cam.get(cv2.CAP_PROP_FRAME_WIDTH))
frame_height = int(cam.get(cv2.CAP_PROP_FRAME_HEIGHT))

video_path = 'C:/Users/Robyn/Desktop/Research Project/CV_Captures/output.mp4'
fourcc = cv2.VideoWriter_fourcc(*'mp4v')
out = cv2.VideoWriter(video_path, fourcc, 20.0, (frame_width, frame_height))

while True:
    ret, frame = cam.read()

    out.write(frame)

    cv2.imshow('Camera', frame)

    if cv2.waitKey(1) == ord('q'):
        break

cam.release()
out.release()
cv2.destroyAllWindows()
```
